```python
# 06. Products - Final Validation

print("PRODUCTS VALIDATION")

# 1. Column Names
print("\n--- Column Names ---")
print(products.columns.tolist())


# 2. Data Types
print("\n--- Data Types ---")
print(products.dtypes)


# 3. Product Category Distribution
print("\n--- Product Category Distribution ---")
print(products["product_category_name"].value_counts().head(15))


# 4. Numeric Columns Summary
print("\n--- Numeric Summary ---")
print(products.describe())


# 5. Invalid Values Check
print("\n--- Invalid Values ---")

print("Negative product weight:",
      (products["product_weight_g"] < 0).sum())

print("Zero product weight:",
      (products["product_weight_g"] == 0).sum())

print("Negative product length:",
      (products["product_length_cm"] < 0).sum())

print("Zero product length:",
      (products["product_length_cm"] == 0).sum())

print("Negative product height:",
      (products["product_height_cm"] < 0).sum())

print("Zero product height:",
      (products["product_height_cm"] == 0).sum())

print("Negative product width:",
      (products["product_width_cm"] < 0).sum())

print("Zero product width:",
      (products["product_width_cm"] == 0).sum())


# 6. Product Photos Validation
print("\n--- Product Photos Validation ---")
print("Minimum photos:",
      products["product_photos_qty"].min())

print("Maximum photos:",
      products["product_photos_qty"].max())

print("Zero photos:",
      (products["product_photos_qty"] == 0).sum())


# 7. Product Name & Description Length Validation
print("\n--- Text Length Validation ---")

print("Product name length <= 0:",
      (products["product_name_lenght"] <= 0).sum())

print("Product description length <= 0:",
      (products["product_description_lenght"] <= 0).sum())


# 8. Missing Values Summary
print("\n--- Missing Values ---")
print(products.isna().sum())


# 9. Duplicate Rows
print("\n--- Duplicate Rows ---")
print("Duplicate rows:", products.duplicated().sum())


# 10. Primary Key Validation
print("\n--- Primary Key Validation ---")
print("Missing product_id:",
      products["product_id"].isna().sum())

print("Unique product_id:",
      products["product_id"].nunique())

print("Total rows:",
      len(products))

print("Duplicate product_id:",
      products["product_id"].duplicated().sum())


# 11. Final Data Quality Status
print("\n========== FINAL STATUS ==========")

if (
    products["product_id"].isna().sum() == 0
    and products["product_id"].duplicated().sum() == 0
    and products.duplicated().sum() == 0
):
    print("Product ID / Duplicate Check: PASS")
else:
    print("Product ID / Duplicate Check: REVIEW")
```
